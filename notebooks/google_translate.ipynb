## Imports

```python
from googletrans import Translator
```

```python
translator = Translator()
```

```python
text = "اصرار حکومت به اسرائیل‌ستیزی چقدر دغدغه مردم ایران است؟"
trans = translator.translate(text, dest='en').text
print(trans)
```
